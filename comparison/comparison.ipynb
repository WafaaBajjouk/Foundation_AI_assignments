# Comparison of the three methods
Reads the results of the three notebooks and compares them on the shared puzzles (N = 3…7, seeds 0–2).
For each case: solved runs out of the total and the median time.

```python
import pandas as pd
```

```python
bt = pd.read_csv("../constraint_propagation_backtracking/results_alg1.csv")
sa = pd.read_csv("../simulated_annealing/results_sa.csv")
ga = pd.read_csv("../genetic_algorithm/results_ga.csv")
```

```python
#backtracking: best setting and only the 3 puzzles shared with SA and GA
bt = bt[(bt["setting"] == "ac + lcv") & (bt["seed"] <= 2)]
sa = sa[sa["N"] <= 7]
```

## Summary

```python
def summarise(df):
    rows = []
    for (frac, n), g in df.groupby(["givens", "N"]):
        k, m = int(g["solved"].sum()), len(g)
        rows.append({"givens": frac, "N": n,
                     "solved": f"{k}/{m}",
                     "median time": round(g["time"].median(), 3)})
    return pd.DataFrame(rows)
```

## Backtracking (AC + LCV)

```python
summarise(bt)
```

## Simulated annealing

```python
summarise(sa)
```

## Genetic algorithm

```python
summarise(ga)
```

## Simulated annealing, N = 8…10
Sizes tested only for SA.

```python
sa_all = pd.read_csv("../simulated_annealing/results_sa.csv")
summarise(sa_all[sa_all["N"] >= 8])
```
